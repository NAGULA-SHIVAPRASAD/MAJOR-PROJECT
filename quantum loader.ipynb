```python
import numpy as np
from tensorflow.keras.models import load_model
from tensorflow.keras.preprocessing import image

#  14 disease names 
disease_names = [
    "Atelectasis", "Cardiomegaly", "Effusion", "Infiltration",
    "Mass", "Nodule", "Pneumonia", "Pneumothorax",
    "Consolidation", "Edema", "Emphysema", "Fibrosis",
    "Pleural_Thickening", "Hernia"
]

#  Load trained model (NO training again)
model = load_model("my_model.keras")

#  New image path
img_path =  r"D:\quantum dataset\images\00000121_004.png"   # <-- change this
#"D:\quantum dataset\images\00000121_004.png"
#  Load and preprocess image
img = image.load_img(img_path, target_size=(224, 224))
img_array = image.img_to_array(img)
img_array = np.expand_dims(img_array, axis=0)
img_array = img_array / 255.0

#  Predict
pred = model.predict(img_array)[0]   # shape (14,)

#  
threshold = 0.75

print("\n Disease Prediction Results:\n")
found = False

for name, prob in zip(disease_names, pred):
    status = " Present" if prob >= threshold else " Not Present"
    print(f"{name:20s} :   --> {status}")
    if prob >= threshold:
        found = True

#  
print("\n===============================")
if found:
    detected = [name for name, prob in zip(disease_names, pred) if prob >= threshold]
    print(" Final Detected Diseases:", detected)
else:
    print(" No disease detected above threshold.")
print("===============================\n")
```
